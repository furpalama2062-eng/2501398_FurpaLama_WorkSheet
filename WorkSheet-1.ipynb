```python
import numpy as np

arr_empty = np.empty((2, 2))
print("1. Empty array (2x2):")
print(arr_empty)
print()

arr_ones = np.ones((4, 2))
print("2. All ones array (4x2):")
print(arr_ones)
print()

arr_full = np.full((3, 3), 7, dtype=int)
print("3. Full array with fill value 7:")
print(arr_full)
print()

given_array = np.array([[1, 2], [3, 4]])
arr_zeros_like = np.zeros_like(given_array)
print("4. Zeros array with same shape and type as given array:")
print(arr_zeros_like)
print()

arr_ones_like = np.ones_like(given_array)
print("5. Ones array with same shape and type as given array:")
print(arr_ones_like)
print()

new_list = [1, 2, 3, 4]
np_array = np.array(new_list)
print("6. NumPy array from list:")
print(np_array)
```

```python
x = np.array([[1, 2], [3, 5]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])

print("x + y =\n", x + y)
print()

print("x - y =\n", x - y)
print()

print("x * 3 =\n", x * 3)
print()

print("x^2 =\n", x ** 2)
print()

print("Dot product of v and w:", np.dot(v, w))
print("Dot product of x and v:", np.dot(x, v))
print("Dot product of x and y:\n", np.dot(x, y))
print()

print("Concatenate x and y along row (axis=0):\n", np.concatenate((x, y), axis=0))
print("Concatenate v and w along column (as 2D arrays):\n", np.column_stack((v, w)))
print()

try:
    print(np.concatenate((x, v), axis=0))
except ValueError as e:
    print("Error:", e)
    print("Explanation: x is 2x2 and v is 1D array of size 2. They do not have the same number of dimensions, so direct concatenation fails.")
```

```python
import numpy as np

A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

I = np.dot(A, np.linalg.inv(A))
print("A * A^-1 =\n", I)
print()

AB = np.dot(A, B)
BA = np.dot(B, A)
print("AB =\n", AB)
print("BA =\n", BA)
print("AB == BA:", np.array_equal(AB, BA))
print()

print("(AB)^T =\n", (AB).T)
print("B^T * A^T =\n", np.dot(B.T, A.T))
print("Check equality:", np.array_equal((AB).T, np.dot(B.T, A.T)))
print()

coeff = np.array([[2, -3, 1], [1, -1, 2], [3, 1, -1]])
const = np.array([[-1], [-3], [9]])

X = np.dot(np.linalg.inv(coeff), const)
print("Solution using inverse method:\n", X)
```
